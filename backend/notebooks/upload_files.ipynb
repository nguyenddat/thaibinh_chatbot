```python
import os
import requests
```

```python
print(os.getcwd())
```

```python
folder_path = "../artifacts/json/"
url = "https://chatbot-tb-back.ript.vn/api/procedure/"

json_files = [f for f in os.listdir(folder_path) if f.endswith(".json")]
json_files
```

```python
files_payload = []
for file_name in json_files:
    file_path = os.path.join(folder_path, file_name)
    files_payload.append(
        ('files', (file_name, open(file_path, 'rb'), 'application/json'))
    )

response = requests.post(url, files=files_payload)
```

```python
try:
    print(response.status_code)
    print(response.json())
except Exception as e:
    print("Lỗi khi parse response:", e)
    print(response.text)
```
