```python
from gtts import gTTS
from pydub import AudioSegment

text = "Căn cứ pháp lý chi tiết của cấp giấy xác nhận thân nhân của người có công?"

# Tạo MP3 từ gTTS
tts = gTTS(text=text, lang='vi')
tts.save("temp.mp3")

# Chuyển sang WAV chuẩn
sound = AudioSegment.from_mp3("temp.mp3")

# Mono 16kHz PCM, đồng thời tăng volume để nghe rõ
sound = sound.set_channels(1).set_frame_rate(16000)
sound = sound + 10  # tăng 10dB, có thể điều chỉnh

# Xuất file WAV
sound.export("can_cu_phap_ly_chi_tiet.wav", format="wav", codec="pcm_s16le")
print("Đã tạo file WAV chuẩn, âm lượng nghe được")
```
